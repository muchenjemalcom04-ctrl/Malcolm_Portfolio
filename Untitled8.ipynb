```python
import yfinance as yf  #source of data 
import numpy as np     # complex math 
import seaborn as sn   #interactive visuals 
import pandas as pd    # clean my data 
import pandas_ta as pdt  # technical analysis 
import matplotlib.pyplot as plt  #statistical visuals 
```

```python
tickers=["0097.KL","0215.KL","1155.KL","5347.KL","5184.KL","1295.KL","5227.KL","6742.KL","2089.KL"]
benchmark_ticker="^KLSE"  #FTSE BURSA MALAYSIA INDEX SET as bench mark to allow portfolio vs malaysia market comparison i.e CAPM
#in Zim's case, bechmark would be the ZSE index 

data=yf.download(tickers + [benchmark_ticker], start="2020-01-01",end="2024-12-31")["Close"] #4yp for continuity purposes[metrics] #+ to combine the lists
data.dropna(inplace=True)
```

```python
print(data.head())
```

```python
benchmark=data[benchmark_ticker] #index 
stocks=data[tickers] # daily stock prices 
stocks 
```

**HANDLING RETURNS**

```python
returns= stocks.pct_change().dropna()
benchmark_returns=benchmark.pct_change().dropna()
returns=returns.loc[benchmark_returns.index] #loc selects returns for the period in which benchmark_returns were present
rf=0.03    #risk free rate 


annual_returns=returns.mean()*252
annual_volatility= returns.std()*np.sqrt(252)
```

**BETA CALCULATION:cov[market,stock]/var(market)**

```python
beta={}
for stock in tickers:
    cov=np.cov(returns[stock],benchmark_returns)[0][1]
    var=np.var(benchmark_returns)
    beta[stock]=cov/var
beta=pd.Series(beta)
beta # risk levels relative to market per each stock. 
#beta = 1 means that market risk=stock risk 
```

**PORTFOLIO METRICS**

```python
sharpe_ratio=(annual_returns-rf)/annual_volatility  #shows the risk adjusted return. Are we making a premium on risk[>0] or are we lossing [<0] #useful for unsystematic risk i.e inflation, war , drought  

treynor_ratio=(annual_returns-rf)/beta 
#how much return obtained per unit of systematic risk[market risks] i.e supply chaoin shifts, demand shifts and COGS 

#Jensen alpha 
market_return=benchmark_returns.mean()*252 
jensen_alpha=annual_returns -(rf+beta*(market_return-rf))

#Max Drawdown: Loss in value from peak to trough 
rolling_max=stocks.cummax()
drawdown=(stocks-rolling_max)/rolling_max
max_drawdown=drawdown.min()

#Historical VaR at 95% confidence 
VaR_95= returns.quantile(0.05)
#VaR is the value at risk, meaning that it is the potential max loss the portfolio will face. CI just gives us insight on how strong our belief is.

```

```python
#Combine all metrics 
metrics=pd.DataFrame({
    "Annual Returns":annual_returns,
    "Annual Volatility": annual_volatility,
    "Sharpe Ratio":sharpe_ratio,
    "Treynor Ratio": treynor_ratio,
    "Jensen Alpha": jensen_alpha,
    "Max Drwadown": max_drawdown,
    "VaR (95%)":VaR_95,
    "Beta":beta
}).round(4)

print("Perfomance Metrics (2020-2024):")
display(metrics)
```
