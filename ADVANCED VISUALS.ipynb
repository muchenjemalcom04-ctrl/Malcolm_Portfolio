```python
# pip install matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
from matplotlib.lines import Line2D

fig, ax = plt.subplots(figsize=(10, 12))
ax.set_xlim(0, 10)
ax.set_ylim(0, 18)
ax.axis("off")

# Colors
BOX = "#F7F9FC"
EDGE = "#6B7280"
TXT  = "#111827"
TAG_IAAS = "#2563EB"
TAG_PAAS = "#059669"
TAG_ML   = "#CA8A04"

def draw_box(x, y, w, h, title, tags, details):
    box = FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.02,rounding_size=8",
                         linewidth=1, edgecolor="#2B2F36", facecolor=BOX)
    ax.add_patch(box)
    ax.text(x+0.3, y+h-0.6, title, ha="left", va="top", fontsize=11, color=TXT, fontweight="bold")
    # tags
    tx = x+0.3
    for label, color in tags:
        ax.text(tx, y+h-1.4, label, fontsize=9, color=color, fontweight="bold")
        tx += len(label)*0.12 + 1.2
    # details
    dy = 2.2
    for d in details:
        ax.text(x+0.3, y+h-dy, f"• {d}", ha="left", va="top", fontsize=9, color="#374151")
        dy += 0.8

def arrow(x1, y1, x2, y2, text=None, dashed=False):
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle="->", color=EDGE, lw=1.5, linestyle="--" if dashed else "-"))
    if text:
        ax.text((x1+x2)/2, (y1+y2)/2 + 0.3, text, fontsize=9, color="#6B7280", ha="center")

steps = [
    ("Data collection & ingestion",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["APIs, ETL, object storage", "Preprocessing and normalization"]),
    ("Underwriting risk assessment",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["Real-time risk & pricing models", "Fraud probability scoring"]),
    ("Policy issuance",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["Workflow + document generation", "Personalized coverage recommendations"]),
    ("Data storage & monitoring",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["Encrypted long-term storage", "Portfolio analytics & model tuning"]),
    ("Claims submission",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["Web/mobile intake", "Docs, photos, telematics uploads"]),
    ("Claims processing & decision",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["ML triage & straight-through approval", "Human review for high-risk flags"]),
    ("Payment & financial compensation",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["Bank/e-wallet payouts", "Audit trails & anomaly checks"]),
    ("Feedback loop",
     [("IaaS", TAG_IAAS), ("PaaS", TAG_PAAS), ("ML", TAG_ML)],
     ["Outcomes → retraining", "Model performance monitoring"]),
]

# Layout boxes vertically
x, y, w, h = 1, 16, 8, 2.8
positions = []
for title, tags, details in steps:
    draw_box(x, y, w, h, title, tags, details)
    positions.append((x, y, w, h))
    y -= 2.5

# Arrows
for i in range(len(positions)-2):
    x1, y1, w1, h1 = positions[i]
    x2, y2, w2, h2 = positions[i+1]
    arrow(x1+w1/2, y1, x2+w2/2, y2+h2)

# Payout → Feedback → Underwriting dashed loop
x1, y1, w1, h1 = positions[-2]  # payout
x2, y2, w2, h2 = positions[-1]  # feedback
arrow(x1+w1/2, y1, x2+w2/2, y2+h2)
# feedback back to underwriting (curved via two segments)
ux, uy, uw, uh = positions[1]   # underwriting
arrow(x2+w2/2, y2, ux+w/2, uy+uh+0.5, text="continuous improvement", dashed=True)

plt.tight_layout()
plt.savefig("vsure_flow.png", dpi=200, bbox_inches="tight")
print("Saved to vsure_flow.png")
```
