```python
try:
    classes=int(input("Enter total classes:"))
    while classes<0:
        print("Invalid! Try again")
        classes=int(input("Enter total classes again: "))
    
    absent=int(input("Enter total for absent: "))
    while absent>classes :
        print("Sorry, try again")
        absent=int(input("Enter total for absent again: "))
    
    attendance=(((classes-absent)/classes)*100)
    if attendance>=75 :
        print("eligible for exam")
    elif attendance<75 :
        print("not eligible. please refer to admin block")
    else:
        print("continuous assessment")                               #to catch an error apply try and except arguement later on ,# absent purposely left open to catch error in try/except arguement 
except:
    print("zero division error, check absent again")
```
