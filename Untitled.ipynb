```python
systolic=int(input("Enter systolic value: "))
while systolic<0 or systolic>250:
    print("invalid systolic value, try again!")
    systolic=int(input("Enter systolic value again: "))
diastolic=int(input("Enter diastolic value: "))
while diastolic<0 or diastolic >180:
    print("invalid diastolic value, try again!")
    diastolic=int(input("Enter diastolic value again: "))
while diastolic>systolic:
    print("invalid diastolic input")
    diastolic=int(input("Enter valid diastolic: "))
Blood_pressure=(systolic/diastolic) 
if Blood_pressure<(120/80):
    print("low risk") 
elif (120/80)<=Blood_pressure<=(139/89):
    print("medium risk")
elif Blood_pressure>(139/89):
    print("high risk")
else:
    print("invalid category")
```
